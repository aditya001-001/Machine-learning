```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from sklearn.model_selection import train_test_split
```

```python
df=pd.read_csv(f"D:\ML\datasets\weatherHistory.csv")
```

```python
df=df.select_dtypes(include='number')
```

```python
df.corr()
```

```python
plt.scatter(df['Temperature (C)'],df['Humidity'])
```

```python
plt.scatter(df['Temperature (C)'],df['Apparent Temperature (C)'])
```

```python
plt.scatter(df['Humidity'],df['Apparent Temperature (C)'])
```

```python
x=df[['Humidity']]
y=df['Apparent Temperature (C)']
```

```python
model=LinearRegression()
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.2,random_state=42)
model.fit(xtrain,ytrain)
```

```python
y_pred = model.predict(xtest)
```

```python
model.score(xtrain,ytrain)=0.3641842095742762
```

```python
model.score(xtest,ytest)=0.3587532329296208
```

```python
plt.scatter(xtest, ytest, color='blue', alpha=0.3, label='Actual Data')
plt.plot(xtest, y_pred, color='red', linewidth=2, label='Regression Line')
plt.xlabel('Humidity')
plt.ylabel('Apparent Temperature (C)')
plt.legend()
plt.show()
```

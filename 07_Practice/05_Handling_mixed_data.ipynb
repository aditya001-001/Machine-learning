# Handling Data and Time based columns

import pandas as pd
import numpy as np

```python


data = {
    "Customer": ["C001", "C002", "C003", "C004", "C005",
                 "C006", "C007", "C008", "C009", "C010"],

    "Order_Date": [
        "2026-01-15",
        "15/02/2026",
        "March 10, 2026",
        "2026/04/22",
        "25-05-2026",
        "06/15/2026",
        "2026-07-18",
        "19-Aug-2026",
        "09/25/2026",
        "2026-10-01"
    ],

    "Order_Time": [
        "10:30 AM",
        "14:45",
        "06:20 PM",
        "09:15 AM",
        "18:30",
        "11:45 AM",
        "20:10",
        "07:30 AM",
        "16:25",
        "09:50 PM"
    ],

    "Amount": [
        1250, 850, 2300, 450, 1750,
        920, 3100, 650, 1450, 2750
    ]
}

df = pd.DataFrame(data)

df
```

```python
df.info()
```

```python
df['Order_Date']=pd.to_datetime(df['Order_Date'],format='mixed')
```

```python
df.info()
```

```python
df['date_year']=df['Order_Date'].dt.year
df.head()
```

```python
df['date_month']=df['Order_Date'].dt.month
df['date_month_name']=df['Order_Date'].dt.month_name()
df.head()
```

```python
df['date_day']=df['Order_Date'].dt.day
df['date_day_no']=df['Order_Date'].dt.day_of_week
df['date_day_name']=df['Order_Date'].dt.day_name()
```

```python
df.head()
```

```python
# is a weekend
df['date_is_weekend']=np.where(df['date_day_name'].isin(['Sunday','Saturday']),1,0)
```

```python
df.head()
```

```python
# I think week has removed from dt class
df['date_week']=df['Order_Date'].dt.week
df.head()
```

```python
df['quarter']=df['Order_Date'].dt.quarter
df.head()
```

```python
df['semester']=np.where(df['quarter'].isin([1,2]),1,2)
```

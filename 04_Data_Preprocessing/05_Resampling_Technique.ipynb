```python
import pandas as pd
from imblearn.over_sampling import SMOTE
from imblearn.under_sampling import RandomUnderSampler
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from imblearn.pipeline import Pipeline
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
df=pd.read_csv(f"D:\ML\datasets\Heart_Disease_Imbalanced.csv")
```

```python
x=df.drop(columns='Heart_Disease')
y=df['Heart_Disease']
cat_col=x.select_dtypes(include='object').columns
num_col=x.select_dtypes(include='number').columns
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
ytrain.value_counts()
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        (
            'num',
            Pipeline([
                ('imputer', SimpleImputer(strategy='median')),
                ('scaler', StandardScaler())
            ]),
            num_col
        ),

        (
            'cat',
            Pipeline([
                ('imputer', SimpleImputer(strategy='most_frequent')),
                ('encoder', OneHotEncoder(handle_unknown='ignore'))
            ]),
            cat_col
        )
    ]
)
```

```python
pipe = Pipeline([
    ('preprocessor', preprocessor),

    ('under', RandomUnderSampler(
        random_state=42,
        sampling_strategy=0.1
    )),

    ('smote', SMOTE(random_state=42)),

    ('model', LogisticRegression())
])
```

```python
pipe.fit(xtrain,ytrain)
```

```python
ytrain.value_counts()
# It happens internally so we can not see it here
```

```python
X_processed = preprocessor.fit_transform(xtrain)

rus = RandomUnderSampler(random_state=42, sampling_strategy=0.1)
X_under, y_under = rus.fit_resample(X_processed, ytrain)

print("After UnderSampling:")
print(pd.Series(y_under).value_counts())

smote = SMOTE(random_state=42)
X_final, y_final = smote.fit_resample(X_under, y_under)

print("\nAfter SMOTE:")
print(pd.Series(y_final).value_counts())
```

```python
ypred = pipe.predict(xtest)
accuracy_score(ytest, ypred)
```

```python
from sklearn.metrics import classification_report

print(classification_report(ytest, ypred))
```
